### Import Libraries:

### Linear Regression Model

```python
x = df["First-year grade"].values
y = df["Second-yare grade"].values
```

```python
print("x:",x)
print("y:",y)
```

```python
w = np.sum((x - np.mean(x)) * (y - np.mean(y))) / np.sum((x - np.mean(x)) ** 2)

b = np.mean(y) - w * np.mean(x)

print("Slope (w):", w)
print("Intercept (b):", b)
```

```python
def linear_regression(x):
    return b + (w * x)
```

### Cost Function

```python
y_pred = linear_regression(x)
errors = y - y_pred
```

```python
def cost_function(y, y_pred):
    m = len(y)
    cost = np.sum((y_pred - y) ** 2) / (2 * m)
    return cost
```

```python
j = cost_function(y, y_pred)
```

### Final Regression Line

```python
sorted_indices = np.argsort(x)

plt.scatter(x, y, label="Actual Data")

plt.plot(x[sorted_indices], y_pred[sorted_indices],
         color="red", label="Regression Line")

plt.xlabel("First-year grade")
plt.ylabel("Second-year grade")
plt.title("Linear Regression")
plt.legend()
plt.grid()

plt.show()
```

### Prediction

```python
# Enter a new first-year grade
new_grade = 2

# Predict the second-year grade
predicted_grade = linear_regression(new_grade)

# Display the prediction
print("Predicted second-year grade:", predicted_grade)
```
